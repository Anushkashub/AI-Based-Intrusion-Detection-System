```python
import os

DATA_PATH = r"D:\AI_Based_Intrusion_Detecction_System\MachineLearningCSV\MachineLearningCVE"

print("Dataset files:")
print(os.listdir(DATA_PATH))
```

```python
import pandas as pd
import os

file_path = os.path.join(
    DATA_PATH,
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"
)

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Traffic labels:")
print(df[" Label"].value_counts())
```

```python
df.columns = df.columns.str.strip()

print("Column names cleaned successfully!")
print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
missing_rows = df[df.isnull().any(axis=1)]

print("Rows containing missing values:")
print(missing_rows)
```

```python
print("Missing values by column:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df = df.drop_duplicates()

print("Duplicate records removed successfully!")
print("New dataset shape:", df.shape)
print("Remaining duplicate records:", df.duplicated().sum())
```

```python
  print("Final dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
```

```python
import numpy as np

infinite_values = np.isinf(df.select_dtypes(include=np.number)).sum().sum()

print("Total infinite values:", infinite_values)
```

```python
if infinite_values > 0:
    print("Infinite values found.")
else:
    print("No infinite values found.")
```

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)

print("Infinite values handled.")
print("Missing values now:", df.isnull().sum().sum())
```

```python
df.dropna(inplace=True)

print("Final cleaning completed!")
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
```

```python
print("Final traffic labels:")
print(df["Label"].value_counts())
```

```python
print("========== FINAL DATASET CHECK ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
print("Infinite values:", np.isinf(df.select_dtypes(include=np.number)).sum().sum())
```
