```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Libraries imported successfully!")
```

```python
# Load preprocessed training and testing data

X_train_scaled = np.load("X_train_scaled.npy")
X_test_scaled = np.load("X_test_scaled.npy")
y_train = np.load("y_train.npy")
y_test = np.load("y_test.npy")

print("Data loaded successfully!")

print("\nX_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Create Logistic Regression model

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print("Logistic Regression model created successfully!")
```

```python
# Train Logistic Regression model

print("Training Logistic Regression model...")
print("This may take some time because the dataset is large.")

lr_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Make predictions using Logistic Regression

lr_predictions = lr_model.predict(X_test_scaled)

print("Logistic Regression predictions generated successfully!")
print("Number of predictions:", len(lr_predictions))
```

```python
# Calculate Logistic Regression evaluation metrics

lr_accuracy = accuracy_score(y_test, lr_predictions)
lr_precision = precision_score(
    y_test,
    lr_predictions,
    average="weighted",
    zero_division=0
)
lr_recall = recall_score(
    y_test,
    lr_predictions,
    average="weighted",
    zero_division=0
)
lr_f1 = f1_score(
    y_test,
    lr_predictions,
    average="weighted",
    zero_division=0
)

print("Logistic Regression Evaluation:")
print("--------------------------------")
print("Accuracy :", lr_accuracy)
print("Precision:", lr_precision)
print("Recall   :", lr_recall)
print("F1-Score :", lr_f1)
```
